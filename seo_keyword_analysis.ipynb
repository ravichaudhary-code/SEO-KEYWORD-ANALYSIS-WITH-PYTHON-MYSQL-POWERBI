```python
import pandas as pd

df = pd.read_csv("seo_keyword_analysis.csv")

df.head()
df.columns
df.shape
```

```python
# check duplicate keywords
print("Duplicate Keywords:", df["Keyword"].duplicated().sum())

# remove duplicates based on Keyword
df = df.drop_duplicates(subset=["Keyword"])

# check new shape
print("New Shape:", df.shape)
```

```python
import re

def clean_keyword(k):
    k = str(k).lower()
    k = re.sub(r'[^a-z0-9\s]', ' ', k)
    k = re.sub(r'\s+', ' ', k)
    return k.strip()

df["Keyword"] = df["Keyword"].apply(clean_keyword)
```

```python
def assign_group(k):
    import re
    
    k = str(k).lower()
    k = re.sub(r'[^a-z0-9\s]', ' ', k)

    # 1 Project Management
    if any(x in k for x in ["pmp","project","management","pmi"]):
        return 1, "Project Management"
    
    # 2 Agile
    elif any(x in k for x in ["scrum","agile"]):
        return 2, "Scrum & Agile"
    
    # 3 Data Analytics
    elif any(x in k for x in [
        "data","analytics","analyst","analysis",
        "statistics","data science","machine learning"
    ]):
        return 3, "Data Analytics"
    
    # 4 Business Analyst
    elif "business analyst" in k:
        return 4, "Business Analyst"
    
    # 5 Six Sigma
    elif "sigma" in k:
        return 5, "Six Sigma"
    
    # 6 Digital Marketing
    elif any(x in k for x in ["marketing","seo","ads","affiliate"]):
        return 6, "Digital Marketing"
    
    # 7 Cyber Security
    elif any(x in k for x in [
        "security","cyber","hacker","cissp","encryption"
    ]):
        return 7, "Cyber Security"
    
    # 8 Python
    elif "python" in k:
        return 8, "Python"
    
    # 9 Excel
    elif "excel" in k:
        return 9, "Excel"
    
    # 10 SQL & Database
    elif any(x in k for x in [
        "sql","database","mysql","dbms","query"
    ]):
        return 10, "SQL & Database"
    
    # 11 Programming
    elif any(x in k for x in [
        "java","javascript","c++","c#","coding","programming",
        "array","function","string","loop","algorithm"
    ]):
        return 11, "Programming"
    
    # 12 AI / ML
    elif any(x in k for x in [
        "ai","artificial intelligence","machine learning","chatgpt"
    ]):
        return 12, "AI/ML"
    
    # 13 Cloud
    elif any(x in k for x in [
        "aws","azure","cloud","devops","docker"
    ]):
        return 13, "Cloud"
    
    # 14 Career
    elif any(x in k for x in [
        "interview","questions","job","salary",
        "resume","career","fresher","experience"
    ]):
        return 14, "Career"
    
    # 🔥 Generic learning keywords
    elif any(x in k for x in [
        "what","how","types","examples","definition",
        "meaning","uses","introduction","benefits"
    ]):
        return 3, "Data Analytics"
    
    # 🔥 General tech keywords
    elif any(x in k for x in [
        "computer","technology","software","application",
        "system","internet"
    ]):
        return 11, "Programming"
    
    # 🔥 Single word tech keywords
    elif len(k.split()) == 1 and any(x in k for x in [
        "c","java","python","sql","aws","html","css","js","vba","api"
    ]):
        return 11, "Programming"
    
    # 🔥 Tech signals
    elif any(x in k for x in [
        "code","dev","tool","platform",
        "server","framework","library"
    ]):
        return 11, "Programming"
    
    # 15 Others
    else:
        return 15, "Others"
```

```python
df[["Keyword_ID","Group_Name"]] = df["Keyword"].apply(
    lambda x: pd.Series(assign_group(x))
)

df["Keyword_ID"].value_counts()
```

```python
df["Title"] = df["Keyword"] + " course | Learn Now"
```

```python
main_table = df[[
    "Title",
    "Keyword",
    "Keyword_ID",
    "Position",
    "Previous position"
]]
```

```python
main_table.head()
main_table.shape
```

```python
main_table.to_csv("main_table.csv", index=False)
```

```python
lookup_search = df.groupby("Keyword_ID", as_index=False)["Search Volume"].mean()
```

```python
lookup_traffic = df.groupby("Keyword_ID", as_index=False).agg({
    "Traffic": "mean",
    "Traffic (%)": "mean"
})
```

```python
lookup_keyword = df[["Keyword_ID","Group_Name"]].drop_duplicates()
```

```python
lookup_metrics = df.groupby("Keyword_ID", as_index=False).agg({
    "CPC": "mean",
    "Keyword Difficulty": "mean",
    "Competition": "mean"
})
```

```python
lookup_search.to_csv("lookup_search.csv", index=False)
lookup_traffic.to_csv("lookup_traffic.csv", index=False)
lookup_keyword.to_csv("lookup_keyword.csv", index=False)
lookup_metrics.to_csv("lookup_metrics.csv", index=False)
```
